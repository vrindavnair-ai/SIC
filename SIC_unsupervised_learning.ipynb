```python
from sklearn.datasets import make_blobs, make_moons
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
import matplotlib.pyplot as plt
import pandas as pd

# 1. Generate sample data
X, labels_true = make_blobs(
    n_samples=300,
    n_features=2,
    centers=3,
    cluster_std=5,
    random_state=123
)

# 2. Original data visualization
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.scatter(X[:, 0], X[:, 1], c=labels_true, alpha=0.7)
plt.title('Dataset #1: Original')

# 3. K-Means clustering (k = 3)
kmeans = KMeans(n_clusters=3, random_state=123, n_init=10)
kmeans_labels = kmeans.fit_predict(X)

plt.subplot(2, 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=kmeans_labels, alpha=0.7)
plt.title('K-Means (k=3)')

# 4. Agglomerative Clustering (k = 3)
agg = AgglomerativeClustering(n_clusters=3)
agg_labels = agg.fit_predict(X)

plt.subplot(2, 2, 3)
plt.scatter(X[:, 0], X[:, 1], c=agg_labels, alpha=0.7)
plt.title('Agglomerative Clustering (k=3)')

# 5. DBSCAN clustering
dbscan = DBSCAN(eps=8, min_samples=5)
db_labels = dbscan.fit_predict(X)

plt.subplot(2, 2, 4)
plt.scatter(X[:, 0], X[:, 1], c=db_labels, alpha=0.7)
plt.title('DBSCAN (eps=8, min_samples=5)')

plt.tight_layout()
plt.show()
```

```python

X, labels_true = make_moons(
    n_samples=300,
    noise=0.1,
    random_state=123
)

# 2. Original data visualization
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.scatter(X[:, 0], X[:, 1], c=labels_true, alpha=0.7)
plt.title('Dataset #1: Original')

# 3. K-Means clustering (k = 3)
kmeans = KMeans(n_clusters=3, random_state=123, n_init=10)
kmeans_labels = kmeans.fit_predict(X)

plt.subplot(2, 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=kmeans_labels, alpha=0.7)
plt.title('K-Means (k=3)')

# 4. Agglomerative Clustering (k = 3)
agg = AgglomerativeClustering(n_clusters=3)
agg_labels = agg.fit_predict(X)

plt.subplot(2, 2, 3)
plt.scatter(X[:, 0], X[:, 1], c=agg_labels, alpha=0.7)
plt.title('Agglomerative Clustering (k=3)')

# 5. DBSCAN clustering
dbscan = DBSCAN(eps=8, min_samples=5)
db_labels = dbscan.fit_predict(X)

plt.subplot(2, 2, 4)
plt.scatter(X[:, 0], X[:, 1], c=db_labels, alpha=0.7)
plt.title('DBSCAN (eps=8, min_samples=5)')

plt.tight_layout()
plt.show()
```
